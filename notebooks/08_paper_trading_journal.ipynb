# Stage 8 — Paper ledger and trade journal

**Self-directed academic project.** The simulator is local and educational; it cannot submit broker orders. A simulated order requires a rationale first, a manually entered reference price, and explicit transaction-cost and slippage assumptions. A later review records what happened and what should change.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

from portfolio_research.paper_trading import (
    add_trade_review,
    load_paper_portfolio,
    record_paper_trade,
)

with TemporaryDirectory() as folder:
    ledger = Path(folder) / "lesson.sqlite3"
    trade = record_paper_trade(
        symbol="SPY",
        side="buy",
        quantity=1,
        reference_price=500,
        rationale="Practice order: document the thesis and risks before entry.",
        transaction_cost_bps_per_side=5,
        slippage_bps_per_side=5,
        database_path=ledger,
        initial_cash=10_000,
    )
    add_trade_review(trade.trade_id, "Review whether the original thesis held.", ledger)
    portfolio = load_paper_portfolio(ledger, initial_cash=10_000)
    print("Cash:", round(portfolio.cash, 2))
    print("Holdings:", portfolio.holdings)
    print("Reason:", portfolio.trades[0].rationale)
    print("Review:", portfolio.trades[0].review)
```

The temporary database keeps this lesson from changing the dashboard's local account. Cash includes modeled costs; the example's reference price is not a live quote or a real fill. The simulator rejects unfunded buys, overselling, invalid symbols, and blank rationales.
