# Stage 9 — Validate inputs and verify an audit trail

**Self-directed academic project.** The application records each evaluated signal and position plus paper-trade actions in a local JSON Lines log. Events are linked by SHA-256 hashes so accidental edits are detectable. This is tamper-evident, not tamper-proof: a person able to rewrite the file can recompute the chain, and the log is not encrypted.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

import pandas as pd

from portfolio_research.audit import audit_signal_run, verify_audit_log
from portfolio_research.config import SmaSettings

with TemporaryDirectory() as folder:
    index = pd.bdate_range("2024-01-01", periods=4)
    signals = pd.Series([None, 0.0, 1.0, 1.0], index=index)
    positions = pd.DataFrame({"SPY": [0.0, 0.0, 0.0, 1.0]}, index=index)
    log = Path(folder) / "audit.jsonl"
    run_id = audit_signal_run(
        "SMA crossover", ["SPY"], signals, positions, SmaSettings(2, 3, 5, 5, 1), log
    )
    print("Run:", run_id)
    print("Verified events:", verify_audit_log(log))
```

The dashboard uses `data/audit.jsonl`; its `.gitignore` rule keeps local research logs out of Git. A hash chain helps identify file modification, but does not provide access control, confidentiality, remote backup, or a trusted timestamp. Store the journal and log on a device you control, avoid writing private personal information into the rationale, and do not treat this as a regulated record system.
