# Stage 10 — Writing up results without overstating them

**Self-directed academic project.** This notebook is a report checklist, not a place to manufacture results. The repository's write-up at `docs/research-writeup.md` deliberately records that empirical strategy results are not yet verified. Keep that statement until you run the app, inspect its data diagnostics, and save the exact assumptions and output.

## Before reporting a number

1. Run the automated tests and record the Git commit.
2. Record source/cache state, retrieval timestamp, observed dates and missing values.
3. Record the selected assets, benchmark, fixed strategy parameters, costs, slippage and execution delay.
4. Confirm that metrics cover only the chronological holdout and that no parameter was chosen after inspecting it.
5. Include failures and limitations. A single favorable chart is not a conclusion.

A real download is intentionally not triggered by this notebook. It would depend on live provider availability and should be run and inspected by the project owner.

## Interpretation questions

- Did the strategy beat its benchmark after costs on the held-out dates?
- How large and long was the worst drawdown, and what did the strategy do during it?
- Did performance depend on one asset, one period, or one parameter choice?
- How would survivorship bias, revised data, and human selection of settings affect the conclusion?
- What evidence would be needed before making a stronger claim?

The answer can be that the strategy failed or that the evidence is too weak. The research process—not a positive return—is the deliverable.
