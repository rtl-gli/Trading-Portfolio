# Stage 4 — Performance and risk metrics

**Self-directed academic project.** This notebook uses synthetic daily returns to check what common metrics measure. Synthetic examples make the arithmetic reproducible; they are not market performance.

CAGR annualises the compounded return over calendar time. Sharpe compares average excess daily return with its variability. Maximum drawdown is the worst fall from a prior wealth peak. Volatility annualises daily standard deviation. Historical VaR is the negated empirical lower-tail return: a positive value indicates a loss quantile, while a negative value means that quantile was a gain. Beta measures covariance with the supplied benchmark relative to benchmark variance.

All five can mislead: sample length matters, tail data are sparse, volatility is not the same as permanent loss, and beta/VaR depend on the selected benchmark and sample. None forecasts future returns.

```python
import numpy as np
import pandas as pd

from portfolio_research.metrics import calculate_metrics

dates = pd.bdate_range("2023-01-02", periods=504)
benchmark = pd.Series(np.linspace(-0.02, 0.02, len(dates)), index=dates)
strategy = 0.5 * benchmark + pd.Series(
    np.sin(np.arange(len(dates))) * 0.003, index=dates
)
metrics = calculate_metrics(strategy, benchmark)
metrics
```
